```python
class Node:
    """Basic Node Class implemented in python"""
    def __init__(self, label, data=None, children=[], parents=[]):
        self.label = label
        self.children = []
        self.parents = []
        self.data = data
        self.data_picker = lambda x: None
                
        self.add_children(children)
        self.add_parents(parents)
            
    def add_children(self, children):
        if type(children) is Node:
            return self.add_child(children)
        elif type(children) is list:
            for child in children:
                self.add_child(child)
        else:
            assert "children must be of type list or of type Node"
        
    def add_child(self, child, flag_add_parent=True):
        if type(child) is Node:
            self.children.append(child)
            if flag_add_parent:
                child.add_parent(self, False)
        else:
            assert "child must be of type Node"
    
    def add_parents(self, parents):
        if type(parents) is Node:
            return self.add_parent(parents)
        elif type(parents) is list:
            for parent in parents:
                self.add_parent(parents)
        else:
            assert "parents must be of type list or of type Node"
    
    def add_parent(self, parent, flag_add_child=True):
        if type(parent) is Node:
            self.parents.append(parent)
            if flag_add_child:
                parent.add_child(self, False)
        else:
            assert "parent must be of type Node"
    
    def num_children(self) -> int:
        if self.children is not None:
            return len(self.children)
        return 0
    
    def num_parents(self) -> int:
        if self.parents is not None:
            return len(self.parents)
        return 0
        
    def get_label(self):
        return self.label
    
    def get_child(self, child_number):
        assert type(child_number) is  int, "child_number must be of type int"
        assert child_number >= 0 and child_number < self.num_children(), "child_number must be between 0 and num children - 1"
        if child_number >= 0 and child_number < self.num_children():
            return self.children[child_number]
    
    def get_children(self):
        return self.children
    
    def get_parents(self):
        return self.parents
    
    def set_label_picker(self, function):
        self.data_picker = function
    
    def __repr__(self):
        return str(self)
        
    def __str__(self):
        data = self.data_picker(self.data)
        if data:
            return f"{self.get_label()} - {data}" 
        return f"{self.get_label()}"
        
import bisect
class SortedNode(Node):
    """Sorted Node Class implemented in python"""
    def __init__(self, label, data=None, children=[], parents=[], sorting_function = lambda x: x.get_label()):
        self.sorting_function = sorting_function
        super().__init__(label, data, children, parents)
        
    def add_parent(self, parent, flag_add_child=True):
        if type(parent) is SortedNode:
            bisect.insort(self.parents, parent, key=self.sorting_function)
            if flag_add_child:
                parent.add_child(self, False)
        else:
            assert "parent must be of type Node"

    def add_child(self, child, flag_add_parent=True):
        if type(child) is SortedNode:
            print(child)
            bisect.insort(self.children, child)
            if flag_add_parent:
                child.add_parent(self, False)
        else:
            assert "child must be of type Node"
```
